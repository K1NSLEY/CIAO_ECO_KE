# Laboratório 4 — Problema Real

## Lista de compras com orçamento limitado

O objetivo é escolher produtos para uma compra mensal sem ultrapassar
um orçamento máximo, buscando maximizar a utilidade total dos produtos.

### Representação

Cada produto possui uma variável binária:

- `0`: produto não escolhido;
- `1`: produto escolhido.

### Espaço de busca

Com 10 produtos existem:

`2¹⁰ = 1024`

soluções candidatas.

### Função objetivo

Maximizar a soma das utilidades dos produtos escolhidos.

### Restrição

O custo total deve ser menor ou igual ao orçamento de R$ 80,00.

### Classificação

O problema é uma variação da Mochila 0/1. O espaço de busca cresce
exponencialmente e, portanto, uma enumeração completa pode se tornar
inviável para instâncias grandes.


```python
# ============================================================
# ATIVIDADE 4 - DO ZERO
# Lista de compras com orçamento limitado
# CIAO_ECO_2026
# ============================================================

import random


produtos = [
    "Arroz",
    "Feijao",
    "Leite",
    "Frango",
    "Ovos",
    "Frutas",
    "Verduras",
    "Cafe",
    "Macarrao",
    "Queijo"
]

precos = [
    25.0,
    9.0,
    6.0,
    22.0,
    12.0,
    15.0,
    10.0,
    18.0,
    7.0,
    20.0
]

utilidades = [
    9,
    8,
    7,
    10,
    8,
    9,
    8,
    10,
    6,
    9
]

orcamento = 80.0


def avaliar_solucao(solucao):
    """
    Calcula o custo e a utilidade de uma solução.
    """

    custo = sum(
        precos[i]
        for i in range(len(produtos))
        if solucao[i] == 1
    )

    utilidade = sum(
        utilidades[i]
        for i in range(len(produtos))
        if solucao[i] == 1
    )

    factivel = custo <= orcamento

    return custo, utilidade, factivel


def gerar_solucao_aleatoria():
    """
    Gera uma solução aleatória.
    """

    return [
        random.randint(0, 1)
        for _ in produtos
    ]


random.seed(42)

solucao = gerar_solucao_aleatoria()

custo, utilidade, factivel = avaliar_solucao(solucao)

print("=" * 60)
print("ATIVIDADE 4 - LISTA DE COMPRAS")
print("=" * 60)

print("\nSolucao aleatoria:")

for i, escolhido in enumerate(solucao):

    if escolhido == 1:
        print(
            f"- {produtos[i]} | "
            f"R$ {precos[i]:.2f} | "
            f"utilidade {utilidades[i]}"
        )

print("\nCusto total: R$ {:.2f}".format(custo))
print("Orcamento: R$ {:.2f}".format(orcamento))
print("Utilidade total:", utilidade)

if factivel:
    print("Solucao factivel: SIM")
else:
    print("Solucao factivel: NAO")


print("""
MODELAGEM

Solucao:
vetor binario com um elemento para cada produto.

Espaco de busca:
para 10 produtos existem 2^10 = 1024 combinacoes.

Funcao objetivo:
maximizar a utilidade total dos produtos escolhidos.

Restricao:
o custo total deve ser menor ou igual ao orcamento.

Classificacao:
o problema e uma variacao da mochila 0/1. O espaco de busca
cresce exponencialmente conforme o numero de produtos aumenta.
Por isso, uma busca exaustiva pode se tornar inviavel para
instancias grandes.
""")
```

## Conclusão

O experimento mostra como um problema cotidiano pode ser transformado
em um problema de otimização.

A solução é representada por um vetor binário, possui uma função objetivo
claramente definida e uma restrição de orçamento.

A principal dificuldade computacional está no crescimento do espaço de
busca: cada produto acrescentado dobra aproximadamente o número de
combinações possíveis.
